```python
# NAME: rentalcar.py

# IMPORT STATEMENTS
# (None)

# FUNCTION DEFINITIONS
# 1. Define classification
def classification():
  # Determine car class
  while True:
    # Prompt the user
    car_class = input("Enter classification code (B, S, L): ")
    # Check if car_class equals "B"
    if car_class == "B":
      # Set class type
      full_class = "Budget"
      # Set price
      class_price = 40
      break
    # Check if car_class equals "S"
    elif car_class == "S":
      # Set class type
      full_class = "Standard"
      # Set price
      class_price = 60
      break
    # Check if car_class equals "L"
    elif car_class == "L":
      # Set class type
      full_class = "Luxury"
      # Set price
      class_price = 80
      break
    # Repeat if input is not correct
    print("Invalid input. Note that prompt is case sensitive.")
  # Return class type and price
  return full_class, class_price

# 2. Define full_mileage
def full_mileage():
  # Prompt user for start odometer
  start_odometer = int(input("Start Odometer: "))
  # Prompt user for end odometer
  end_odometer = int(input("End Odometer: "))
  # Calculate the total miles
  total_miles = end_odometer - start_odometer
  # Return the total miles
  return total_miles

# 3. Define charged_mileage
def charged_mileage(total_miles):
  # Determine if the total miles is over 100
  if total_miles > 100:
    # Subtract 100 from the total miles
    charged_miles = total_miles - 100
    # Return charged miles
    return charged_miles
  # If total miles are less that 100 miles are not charged
  else:
    # Set charged miles to 0
    charged_miles = 0
    # Return charged miles
    return charged_miles

# 4. Define days_rented
def days_rented():
  # Prompt user
  days = int(input("Days Rented: "))
  # Return days
  return days

# 5. Define total_cost
  # Multipy class price by days
  # Multiply charged miles by 0.35
  # Add charged days by charged miles
  # Return total charge

# PROGRAM LOGIC
# Call all functions
full_class, class_price = classification()
total_miles = full_mileage()
charged_miles = charged_mileage(total_miles)
days = days_rented()
# Print rental summary
print("--- RENTAL SUMMARY ---")
print("Classification: " + full_class)
print("Total Miles: " + str(total_miles))
print("Charged Miles: " + str(charged_miles))
```
