```python
# IMPORT STATEMENTS
# (None)

# FUNCTION DEFINITIONS
# 1. Define is_palindrome
#   - Make the word case friendly
#   - Reverse the word
#   - Check if the word and reversed word are the same
#   - Return true if both words are the same
def is_palindrome(word):
    word = word.lower()
    backwards_word = word[::-1]
    return word == backwards_word

# PROGRAM LOGIC
# Prompt the user
word = input("Enter a word:")
# Call the function
if is_palindrome(word):
    print("That word is a palindrome.")
else:
    print("That word is not a palindrome.")
```
