```python
# NAME: gradebook.py
# IMPORT STATEMENTS
# None

# FUNCTION DEFINITIONS
# Define add_students
def add_students(gradebook):
  # Loop until user inputs "done"
  while True:
      # Prompt the user for name
      name = input("Enter student name (or 'done'): ")
      # Check if name equals "done"
      if name.lower() == "done":
        # Break loop
        break
      # If name does not equal "done"
      else:
        # Prompt user for grade
        grade = int(input(f"Enter {name}'s grade: "))
        # Add name and grade to gradebook
        gradebook[name] = grade

# Define calculate average
def calculate_average(gradebook):
  # If no names are entered
  if len(gradebook) == 0:
    # Return 0 to prevent divisions of 0
    return 0
  # If any names have been entered
  else:
    # Calculate the sum
    total = sum(gradebook.values())
    # Calculate the average
    average = total / len(gradebook)
  # Return average
  return average

# PROGRAM LOGIC
# Create gradebook directory
gradebook = {}
# Call add_students
add_students(gradebook)
# Print class report
print("--- Class Report ---")
# If no names were entered
if len(gradebook) == 0:
  # Print result
  print("No students in the gradebook.")
# For each student in directory
for student, grade in gradebook.items():
  # Print student's name and grade
  print(f"{student}: {grade}")
# Print the class average
avg = calculate_average(gradebook)
print(f"Average: {avg:.2f}")
```
