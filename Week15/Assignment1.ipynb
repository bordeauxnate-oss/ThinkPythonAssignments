```python
class Task:
    def __init__(self, name, priority):
        self.name = name
        self.priority = priority

    def __str__(self):
        priority_map = {1: "High", 2: "Medium", 3: "Low"}
        return f"[{priority_map.get(self.priority, 'Unknown')}] {self.name}"


class TaskManager:
    def __init__(self, history_file="completed_tasks.txt"):
        self.tasks = []
        self.history_file = history_file

    def add_task(self, name, priority):
        if priority not in [1, 2, 3]:
            print("Invalid priority. Use 1 (High), 2 (Medium), or 3 (Low).")
            return
        task = Task(name, priority)
        self.tasks.append(task)
        print("Task added successfully.\n")

    def list_tasks(self):
        if not self.tasks:
            print("\nNo tasks available.\n")
            return []

        print("\nYour Tasks:")
        sorted_tasks = sorted(self.tasks, key=lambda t: t.priority)

        for idx, task in enumerate(sorted_tasks, start=1):
            print(f"{idx}. {task}")
        print()

        return sorted_tasks

    def complete_task(self, task_index):
        sorted_tasks = sorted(self.tasks, key=lambda t: t.priority)

        if task_index < 1 or task_index > len(sorted_tasks):
            print("Invalid task number.\n")
            return

        task_to_remove = sorted_tasks[task_index - 1]
        self.tasks.remove(task_to_remove)

        # Save completed task to file
        with open(self.history_file, "a") as f:
            f.write(f"{task_to_remove.name}\n")

        print(f"Task '{task_to_remove.name}' completed and saved.\n")


def main():
    manager = TaskManager()

    while True:
        print("=== Personal Task Manager ===")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Complete Task")
        print("4. Exit")

        choice = input("Choose an option: ").strip()

        if choice == "1":
            name = input("Enter task name: ").strip()
            try:
                priority = int(input("Enter priority (1=High, 2=Medium, 3=Low): ").strip())
            except ValueError:
                print("Priority must be a number.\n")
                continue

            manager.add_task(name, priority)

        elif choice == "2":
            manager.list_tasks()

        elif choice == "3":
            tasks = manager.list_tasks()
            if not tasks:
                continue

            try:
                task_num = int(input("Enter task number to complete: ").strip())
            except ValueError:
                print("Please enter a valid number.\n")
                continue

            manager.complete_task(task_num)

        elif choice == "4":
            print("Exiting Task Manager. Goodbye!")
            break

        else:
            print("Invalid choice. Please try again.\n")


if __name__ == "__main__":
    main()
```
