```python
# NAME: rectangle.py
# IMPORT STATEMENTS
# None

# FUNCTION DEFINITIONS
# Define calculate_rectangle
def calculate_rectangle():
  # Calculate area
  area = length * width
  # Calculate perimeter
  perimeter = 2 * (length + width)
  # Return area and perimeter as a tuple
  return area, perimeter

# PROGRAM LOGIC
# Prompt user for length
length = float(input("Enter the length of the rectangle: "))
# Prompt user for width
width = float(input("Enter the width of the rectangle: "))
# Call calculate_rectangle and unpack the tuple
my_area, my_perimeter = calculate_rectangle()
# Print area
print("Area:", my_area)
# Print perimeter
print("Perimeter:", my_perimeter)
```
