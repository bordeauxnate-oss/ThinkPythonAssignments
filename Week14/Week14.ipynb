```python
# NAME: library.py
# IMPORT STATEMENTS
# None

# CLASS DEFINITIONS
# Define Book class
class Book:
  # Initialize (Attributes = Variables)
  def __init__(self, title, author):
    # Title of book
    self.title = title
    # Author of book
    self.author = author
    # Variable to track if book is available
    self.is_available = True

  # Class Methods (Behaviors = Functions)
  def __str__(self):
    # If book is available
    if self.is_available == True:
      # Set status to available
      status = "Available"
    # Otherwise
    else:
      # Set status to checked out
      status = "Checked Out"
    # Return title, author and status
    return f"{self.title} by {self.author} - {status}"

  # Define checkout
  def checkout(self):
    # Set self.is_available to False
    self.is_available = False

  # Define return_book
  def return_book(self):
    # Set self.is_available to True
    self.is_available = True

# PROGRAM LOGIC
# Define book1
book1 = Book("Think Python", "Allen Downey")
# Define book2
book2 = Book("The Hobbit", "J.R.R. Tolkien")

# Print library status
print("--- Library Status ---")
# Print book1
print(book1)
# Print book2
print(book2)

# Checkout book2
print("\nChecking out 'The Hobbit'...\n")
book2.checkout()

# Print updated status
print("--- Updated Status ---")
# Print book1
print(book1)
# Print book2
print(book2)
```
