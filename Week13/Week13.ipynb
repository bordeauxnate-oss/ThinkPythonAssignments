```python
# NAME: journal.py
# IMPORT STATEMENTS
# None

# FUNCTION DEFINITIONS
# Define write_entry
def write_entry(journal, entry):
  # Open file in append mode
  with open(journal, 'a') as file:
    # Write entry followed by new line
    file.write(entry + "\n")

# Define read_journal
def read_journal(file):
  # If file exists
  try:
    # Open file in read mode
    with open(journal, 'r') as file:
      # Read contents
      contents = file.read()
      # Print header
      print("--- Your Journal ---")
      # Print contents
      print(contents)
  # If file does not exist
  except FileNotFoundError:
    # Print error message
    print("File not found.")

# PROGRAM LOGIC
# Set journal filename
journal = "journal.txt"

# Loop until user inputs "Q"
while True:
  # Prompt user
  prompt = input("(W)rite, (R)ead, or (Q)uit: ").upper()

  # If prompt equals "W"
  if prompt == "W":
    # Prompt user for entry
    entry = input("Enter your entry: ")
    # Call write_entry
    write_entry(journal, entry)
    # Print save message
    print("Entry saved.")

  # If prompt equals "R"
  elif prompt == "R":
    # Call read_journal
    read_journal(journal)

  # If prompt equals "Q"
  elif prompt == "Q":
    # Break loop
    break

  # If prompt is invalid
  else:
    # Print error message
    print("Invalid input.")
```
