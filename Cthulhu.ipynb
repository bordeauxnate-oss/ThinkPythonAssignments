```python
import random
import json
import os

SAVE_FILE = 'game_state.json'

def save_game(investigator_data):
  with open(SAVE_FILE, 'w') as f:
    json.dump(investigator_data, f, indent=4)
  print(f"Game state saved to {SAVE_FILE}")

def load_game():
  if os.path.exists(SAVE_FILE):
    with open(SAVE_FILE, 'r') as f:
      return json.load(f)
  else:
    return {
        'Adam Benchley': 'Alive',
        'Ahmed Salah': 'Alive',
        'Borden': 'Alive',
        'Elizabeth Ives': 'Alive',
        'Fatima Safar': 'Alive',
        'John Morgan': 'Alive',
        'Rasputin': 'Alive',
        'Sergant Ian Welles': 'Alive',
        'Sister Beth': 'Alive',
        'The Kid': 'Alive',
        'Adilah Vidal': 'Alive',
        'Alex Parker': 'Alive',
        'Ariele Veturi': 'Alive',
        'Father Luke': 'Alive',
        'Hailia Hovath': 'Alive',
        'Tony Boyle': 'Alive',
        'Margarethe Richter': 'Alive',
        'Mary Diaz': 'Alive',
        'Sam Wood': 'Alive',
        'Sergeant Mario Braghieri': 'Alive',
        'Agatha May': 'Alive',
        'Huikong': 'Alive',
        'Jakub Peters': 'Alive',
        'Julien Dulik': 'Alive',
        'Leon Barnes': 'Alive',
        'Mike Kozlov': 'Alive',
        'Ruth Weber': 'Alive',
        'Sally Withers': 'Alive',
        'Sandra Atkinson': 'Alive',
        'Stella Fontaine': 'Alive',
        'Anika Biko': 'Alive',
        'Bruno Vitale': 'Alive',
        'Gary West': 'Alive',
        'John Fernsby': 'Alive',
        'Sylvia Doyle': 'Alive',
        'Rosario Guerra': 'Alive'
    }

investigator = load_game()

investigator_names = list(investigator.keys())

while True:
  investigator_1 = random.choice(investigator_names)
  if investigator[investigator_1] == "Alive":
    break
while True:
  investigator_2 = random.choice(investigator_names)
  if investigator_1 != investigator_2:
    if investigator[investigator_2] == "Alive":
      break

insanity = ['Catatonia', 'Codependency', 'Fear of Crowds', 'Hypochondria',
            'Kleptomania', 'Masochism', 'Obsessive Disorder', 'Paranoia',
            'Psychotic Outbreaks', 'Pyromania', 'Recurring Trauma',
            'Short-Term Memory Loss', 'Autophobia', 'Bloodlust',
            'Compulsive Accumulating', 'Hallucinations', 'Impulsive Aggression',
            'Megalomainia', 'Overcaring', 'Traveling Fugue',
            'Dissociative Identity', 'Hyperactivity', 'Shared Psychosis',
            'Uncontrollable Fear']

def choose_insanity():
  available_insanity_pool = list(insanity) # Work on a copy

  # Choose first insanity
  player_insanity_1_val = random.choice(available_insanity_pool)
  available_insanity_pool.remove(player_insanity_1_val)

  # Handle Dissociative Identity for player_insanity_1
  if player_insanity_1_val == 'Dissociative Identity':
    if len(available_insanity_pool) < 2:
      player_insanity_1_val = "Dissociative Identity (No sub-insanities chosen due to limited pool)"
    else:
      di_sub_insanity_1 = random.choice(available_insanity_pool)
      available_insanity_pool.remove(di_sub_insanity_1)
      di_sub_insanity_2 = random.choice(available_insanity_pool)
      available_insanity_pool.remove(di_sub_insanity_2)
      player_insanity_1_val = f"Dissociative Identity ({di_sub_insanity_1}, {di_sub_insanity_2})"

  # Choose second insanity, ensuring it's distinct from the first (and its sub-insanities if DI)
  player_insanity_2_val = random.choice(available_insanity_pool)
  available_insanity_pool.remove(player_insanity_2_val)

  # Handle Dissociative Identity for player_insanity_2
  if player_insanity_2_val == 'Dissociative Identity':
    if len(available_insanity_pool) < 2:
      player_insanity_2_val = "Dissociative Identity (No sub-insanities chosen due to limited pool)"
    else:
      di_sub_insanity_1 = random.choice(available_insanity_pool)
      available_insanity_pool.remove(di_sub_insanity_1)
      di_sub_insanity_2 = random.choice(available_insanity_pool)
      available_insanity_pool.remove(di_sub_insanity_2)
      player_insanity_2_val = f"Dissociative Identity ({di_sub_insanity_1}, {di_sub_insanity_2})"

  return player_insanity_1_val, player_insanity_2_val



monster = ['Byakhee', 'Chthonic Entity', 'Deep One', 'Fire Vampire', 'Ghoul',
           'Hunting Horror', 'Shoggoth', 'Bokrug', 'Dimensional Shambler',
           'Elder Thing', 'Flying Polyp', 'Hound of Tindalos', 'Nightgaunt',
           'Star Vampire', 'Tcho-Tcho', 'Yugg', 'Dhole', 'Fisher From Outside',
           'Ghast', 'Gug', 'Gyaa-Yothn', 'Serpent Man', 'The Unnamable',
           'Albino Penguin', 'Ancient One', 'Man of Leng', 'Monstrous Frog',
           'Sand Dweller', 'Spider of Leng', 'White Polypus Thing']

def episode_monsters(season, episode):
  try:
    season = int(season)
    episode = int(episode)
  except ValueError:
    print("Season and episode must be numbers.")
    return None, None

  if season == 1:
    if episode == 1:
      monster_1 = 'Byakhee'
      monster_2 = 'Fire Vampire'
    elif episode == 2:
      monster_1 = 'Ghoul'
      monster_2 = 'Cthonic Entity'
    elif episode == 3:
      monster_1 = 'Deep One'
      monster_2 = 'Hunting Horror'
    elif episode == 4:
      monster_1 = 'Byakhee'
      monster_2 = 'Shoggoth'
    elif episode == 5:
      monster_1 = 'Deep One'
      monster_2 = 'Chthonic Entity'
    elif episode == 6:
      monster_1 = 'Ghoul'
      monster_2 = 'Fire Vampire'
    else:
        return None, None
  elif season == 2:
    if episode == 1:
      monster_1 = 'Tcho-Tcho'
      monster_2 = 'Elder Thing'
    elif episode == 2:
      monster_1 = 'Nightgaunt'
      monster_2 = 'Flying Polyp'
    elif episode == 3:
      monster_1 = 'Ghoul'
      monster_2 = 'Star Vampire'
    elif episode == 4:
      monster_1 = 'Hunting Horror'
      monster_2 = 'Bokrug'
    elif episode == 5:
      monster_1 = 'Hound of Tindalos'
      monster_2 = 'Yugg'
    elif episode == 6:
      monster_1 = 'Tcho-Tcho'
      monster_2 = 'Dimensional Shambler'
    else:
        return None, None
  elif season == 3:
    if episode == 1:
      monster_1 = 'Ghast'
      monster_2 = 'The Unnamable'
    elif episode == 2:
      monster_1 = 'Fisher From Outside'
      monster_2 = 'Dhole'
    elif episode == 3:
      monster_1 = 'Gyaa-Yothn'
      monster_2 = 'Dhole'
    elif episode == 4:
      monster_1 = 'Ghast'
      monster_2 = 'Gug'
    elif episode == 5:
      monster_1 = 'Serpent Man'
      monster_2 = 'Gyaa-Yothn'
    elif episode == 6:
      monster_1 = 'Serpent Man'
      monster_2 = 'Fisher From Outside'
  elif season == 4:
    if episode == 1:
      monster_1 = 'Man of Leng'
      monster_2 = 'Spider of Leng'
    elif episode == 2:
      monster_1 = 'Albino Penguin'
      monster_2 = 'Ancient One'
    elif episode == 3:
      monster_1 = 'Spider of Leng'
      monster_2 = 'White Polypus Thing'
    elif episode == 4:
      monster_1 = 'Man of Leng'
      monster_2 = 'Monstrous Frog'
    elif episode == 5:
      monster_1 = 'Sand Dweller'
      monster_2 = 'White Polypus Thing'
    elif episode == 6:
      monster_1 = 'Sand Dweller'
      monster_2 = 'Albino Penguin'
    else:
        return None, None
  else:
      return None, None

  return monster_1, monster_2

def unknown_monsters(monster_1, monster_2):

  available_monsters = list(monster)

  if monster_1 in available_monsters: available_monsters.remove(monster_1)
  if monster_2 in available_monsters: available_monsters.remove(monster_2)

  if len(available_monsters) < 2:
      if len(available_monsters) == 1:
          unknown_monster1 = random.choice(available_monsters)
          unknown_monster2 = None
      else:
          unknown_monster1 = None
          unknown_monster2 = None
      return unknown_monster1, unknown_monster2

  unknown_monster1 = random.choice(available_monsters)
  available_monsters.remove(unknown_monster1)
  unknown_monster2 = random.choice(available_monsters)

  return unknown_monster1, unknown_monster2

elder_ones = ['Cthulhu', 'Hastur', 'The Black Goat of the Woods', 'Yog-Sothoth',
              'Azathoth', 'Tsathoggua', 'Nyarlathotep']
elder_one = random.choice(elder_ones)

relics = ['Ancient Idol', 'Brass Compass', 'Crystal Ball', 'Jade Amulet',
          'Lucky Coin', 'Mysterious Key', 'Replica Blade', 'Ritual Mask',
          'Ruby Talisman', 'Runecarved Spear', 'Shiny Jewel', 'Sinister Book',
          'Glowing Ring', 'Handheld Magnifier', 'Old Rosary', 'Ritual Dagger',
          'Viol', 'Wooden Pipe']
relic_1 = random.choice(relics)
while True:
  relic_2 = random.choice(relics)
  if relic_1 != relic_2:
    break

season_input = input("Season:")
episode_input = input("Episode:")


episode_monster1, episode_monster2 = episode_monsters(season_input, episode_input)


unknown_monster_result = unknown_monsters(episode_monster1, episode_monster2)

player_insanity_1, player_insanity_2 = choose_insanity()

print(elder_one)
print(f"{investigator_1} ({player_insanity_1}) ({relic_1})")
print(f"{investigator_2} ({player_insanity_2}) ({relic_2})")

filtered_unknown_monsters = [m for m in unknown_monster_result if m is not None]
if filtered_unknown_monsters:
  print(", ".join(filtered_unknown_monsters))
else:
  print("No unknown monsters were found.")
outcome = input("Did you succeed?: ")
if outcome.upper() == "N":
  investigator[investigator_1] = "Dead"
  investigator[investigator_2] = "Dead"

save_game(investigator)

alive_investigators = [status for status in investigator.values() if status == 'Alive']
if not alive_investigators:
  print("All investigators are dead. The game is over!")
```
