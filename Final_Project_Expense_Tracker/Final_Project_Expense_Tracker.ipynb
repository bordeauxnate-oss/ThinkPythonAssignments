```python
# IMPORT STATEMENTS
# None

# FUNCTION DEFINITIONS
# Define sum_all
def sum_all(expenses):
  # Return sum of all expenses
  return sum(exp["amount"] for exp in expenses)

# Define sum_by_category
def sum_by_category(expenses, category):
  # Return sum of all expenses for chosen category
  return sum(exp["amount"] for exp in expenses if exp["category"].lower() == category.lower())

# Define sum_by_vendor
def sum_by_vendor(expenses, vendor):
  # Return sum of all expenses for chosen vendor
  return sum(exp["amount"] for exp in expenses if exp["vendor"].lower() == vendor.lower())

# PROGRAM LOGIC
# Create expenses list
expenses = []

# Create loop
while True:
  # Print menu
  print("\n--- Expense Tracker ---")
  print("1. Add New Expense")
  print("2. Get Totals")
  print("3. Exit")

  # Prompt user
  selection = input("Selection: ")

  # If input equals 1
  if selection == "1":
    # Prompt user for category
    category = input("Enter Category: ")
    # Prompt user for vendor
    vendor = input("Enter Vendor: ")

    # Create loop
    while True:
      # Prompt user for amount
      amount_input = input("Enter Amount: ")
      # If input is valid
      try:
        # Set amount as a float
        amount = float(amount_input)
        # Break loop
        break
      # If input is invalid
      except ValueError:
        # Print error message
        print("Invalid amount. Please enter a number.")

    # Prompt user for date
    date = input("Enter Date: ")

    # Create expense directory
    expense = {
        "category": category,
        "vendor": vendor,
        "amount": amount,
        "date": date
    }

    # Add directory to expenses list
    expenses.append(expense)

  # If input equals 2
  elif selection == "2":

    # Print options menu
    print("\n--- Totaling Options ---")
    print("A: All Expenses")
    print("C: Filter by Category")
    print("V: Filter by Vendor")

    # Prompt user for choice
    choice = input("Choice: ").upper()

    # If input equals A
    if choice == "A":
      # Call sum_all function
      total = sum_all(expenses)
      # Print result
      print(f"Total of all expenses: ${total:.2f}")

    # If input equals C
    elif choice == "C":
      # Prompt user for category
      cat = input("Enter Category to sum: ")
      # Call sum_by_category function
      total = sum_by_category(expenses, cat)
      # Print result
      print(f"The total for {cat} is: ${total:.2f}")

    # If input equals V
    elif choice == "V":
      # Prompt user for vendor
      ven = input("Enter Vendor to sum: ")
      # Call sum_by_vendor function
      total = sum_by_vendor(expenses, ven)
      # Print result
      print(f"The total for {ven} is: ${total:.2f}")

    # If input is invalid
    else:
      # Print error message
      print("Invalid choice.")

  # If input equals 3
  elif selection == "3":
    # Break loop
    break

  # If input is invalid
  else:
    # Print error message
    print("Invalid selection.")
```
